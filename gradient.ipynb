```python
import matplotlib.pyplot as plt
```

```python
def function(x):
    return (x+3) ** 2

def derivative(x):
    return 2 * (x+3)
```

```python
x = 2
learning_rate = 0.1
iterations = 20
```

```python
x_values = [x]
y_values = [function(x)]
```

```python
print("Gradient Descent")
print("----------------")
print("Initial x =", x)
print()
```

```python
for i in range(iterations):
    gradient = derivative(x)
    x = x - learning_rate * gradient

    x_values.append(x)
    y_values.append(function(x))

    print(f"Iteration {i + 1}: x = {x:.6f}, y = {function(x):.6f}")
```

```python
print("\nLocal Minimum:")
print("x =", x)
print("y =", function(x))
```

```python
plt.plot(x_values, y_values, marker="o")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent for y = (x + 3)^2")
plt.scatter(-3, 0, color="blue", s=100, label="Minimum (-3, 0)", zorder=6)
plt.annotate("Minimum (-3, 0)", xy=(-3, 0), xytext=(-5, 5), arrowprops=dict(arrowstyle="->"))
plt.grid()
plt.show()
```
